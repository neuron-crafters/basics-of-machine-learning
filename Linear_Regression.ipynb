```python
# Step 1: Import the tools
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Step 2: Prepare our data
X = [[1], [2], [3], [4], [5]]   # Hours studied (X = input)
y = [35, 50, 65, 75, 85]        # Marks (y = output)
```

```python
# Step 3: Create the model
model = LinearRegression()
```

```python
# Step 4: Teach the model (fit means "learn from data")
model.fit(X, y)
```

```python
# Step 5: Predict marks for 6 hours of study
predicted_marks = model.predict([[6]])
print("Predicted Marks for 6 hours:", predicted_marks[0])
```

```python
# Step 6: Draw the dots and the line
plt.scatter(X, y, color='blue', label='Actual Marks')  # dots
plt.plot(X, model.predict(X), color='red', label='Best Fit Line')  # line
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
